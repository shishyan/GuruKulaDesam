# Continuous AI Batch Video Extension Pipeline (2x Length)
This notebook extends workspace videos (such as the **1,049 clips** in `production/google_flow_renders/` and `production/batch03/`) to **at least double their original length (`8s -> 16s+`)** at `1280x720 @ 24fps` using **Lightricks/LTX-Video** Image-to-Video + FFmpeg Optical-Flow Interpolation.

### How It Works
1. Clones `GuruKulaDesam` directly onto the Colab GPU VM.
2. Extracts the final valid frame (`last_frame.png`) of each `8s-10s` scene clip.
3. Synthesizes `161` forward frames (`~6.7s`) with **LTX-Video** (`bfloat16`) guided by the scene's spiritual prompt.
4. Uses FFmpeg `minterpolate` optical flow + Lanczos `1280x720` upscaling to match the exact duration of the original clip (`8s-10s`) and concatenates them into a seamless `16s-20s` (`2x`) master clip.
5. Automatically commits and pushes extended batches back to GitHub.
